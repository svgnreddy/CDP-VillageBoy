# Comprehensive string example

```python
# Comprehensive string Example - let's Call this notebook
# comprehensive_string_example.ipynb
#
# ============================================
# Pharmacy Prescription Label Processing Example
# Demonstrates 20+ Python string methods
# ============================================

# Raw, messy input – this is how data often looks when it comes to you (Data Analyst/Data scientist) for analysis
# This data may come from a scanned form or manual entry or excel sheets or any other means
raw_entry = "  metformin HYDROCHLORIDE 500mg  "
ndc_code = "00093-1048-01"
patient_name = "Priya Garg"
instructions = "Take 1 tablet twice daily with food"
batch_id = "487"
print("#### STEP 0: printing the raw drug entry (with trailing & beginning spaces  ) ####")
print(f"before strip(): '{raw_entry}'")

print("#### STEP 1: Let's start cleaning the raw drug entry #####")

# 1. strip() -- removes leading/trailing whitespace
cleaned = raw_entry.strip()
print(f"After strip(): '{cleaned}'")

# 2. lstrip() -- removes whitespace from the left only
print(f"lstrip(): '{raw_entry.lstrip()}'")

# 3. rstrip() -- removes whitespace from the right only
print(f"rstrip(): '{raw_entry.rstrip()}'")

# 4. lower() -- converts everything to lowercase
print(f"lower(): {cleaned.lower()}")

# 5. upper() -- converts everything to UPPERCASE (e.g. for a warning label)
print(f"upper(): {cleaned.upper()}")

# 6. title() -- capitalizes each word, good for display formatting
drug_display = cleaned.title()
print(f"title(): {drug_display}")

# 7. swapcase() -- flips upper/lower case, just to demonstrate
print(f"swapcase(): {cleaned.swapcase()}")

print("\n----****** STEP 2: let's now start checking and searching the drug name ----******")

# 8. startswith() -- checks if the string starts with something
print(f"startswith('metformin'): {cleaned.lower().startswith('metformin')}")

# 9. endswith() -- checks if the string ends with something
print(f"endswith('500mg'): {cleaned.lower().endswith('500mg')}")

# 10. find() -- returns the position of a substring (-1 if not found)
position = cleaned.lower().find("hydrochloride")
print(f"find('hydrochloride'): position {position}")

# 11. rfind() -- finds the position of the LAST occurrence
print(f"rfind('m'): position {cleaned.lower().rfind('m')}")

# 12. count() -- counts how many times something appears
print(f"count('m'): {cleaned.lower().count('m')}")

print("\n#### STEP 3: Splitting and rebuilding text ####")

# 13. split() -- breaks the string into a list of words
words = cleaned.split()
print(f"split(): {words}")

# 14. join() -- glues a list back together with a separator
rejoined = "-".join(words)
print(f"join(): {rejoined}")

# 15. rsplit() -- splits from the right, useful for limited splits
ndc_parts = ndc_code.rsplit("-", 1)
print(f"rsplit('-', 1) on NDC: {ndc_parts}")

# 16. partition() -- splits into exactly 3 parts: before, separator, after
before, sep, after = ndc_code.partition("-")
print(f"partition('-'): before='{before}', sep='{sep}', after='{after}'")

print("\n#### STEP 4: Validating characters (useful for form checks) ####")

# 17. isalpha() -- True if ALL characters are letters
print(f"isalpha() on 'metformin': {'metformin'.isalpha()}")

# 18. isdigit() -- True if ALL characters are digits
print(f"isdigit() on '500': {'500'.isdigit()}")

# 19. isalnum() -- True if characters are letters AND/OR digits (no spaces/symbols)
print(f"isalnum() on 'batch487': {'batch487'.isalnum()}")

# 20. isspace() -- True if the string is ONLY whitespace
print(f"isspace() on '   ': {'   '.isspace()}")

print("\n#### STEP 5: Formatting the final printed label ####")

# 21. replace() -- swaps one substring for another
short_name = drug_display.replace("Hydrochloride", "HCl")
print(f"replace(): {short_name}")

# 22. zfill() -- pads a number with leading zeros (e.g. for batch IDs)
padded_batch = batch_id.zfill(6)
print(f"zfill(6): {padded_batch}")

# 23. center() -- centers text within a given width, useful for printed labels
print(f"center(40): '{short_name.center(40, '-')}'")

# 24. format() -- inserts values into a template string
label_template = "{name} | {drug} | Batch: {batch}"
final_label = label_template.format(
    name=patient_name.title(),
    drug=short_name,
    batch=padded_batch
)
print(f"format(): {final_label}")

print("\n**** FINAL PRESCRIPTION LABEL TO BE PRINTED ON THE INVOICE ****")
print(final_label)

```

# Pharmacy Inventory Management Example

```python
# ============================================
# Pharmacy Inventory Management System
# Demonstrates all 11 Python list methods
# Let's call it List_demo_pharm_inventory.ipynb
# ============================================

# Starting inventory of drugs in stock
inventory = ["Aspirin", "Ibuprofen", "Metformin","Tylenol","Mucinex","Zyrtec","Azithromycin"]
print(f"Starting inventory: {inventory}")

print("\n---- Receiving new stock ----")

# 1. append() -- add a single new drug to the end of the list
inventory.append("Lisinopril")
print(f"After append('Lisinopril'): {inventory}")

# 2. extend() -- add multiple new drugs at once (e.g. a bulk delivery)
inventory.extend(["Atorvastatin", "Omeprazole"])
print(f"After extend([...]): {inventory}")

# 3. insert() -- add a drug at a specific position (e.g. priority placement)
inventory.insert(0, "Epinephrine")   # emergency drug placed first
print(f"After insert(0, 'Epinephrine'): {inventory}")

print("\n---- Checking stock details ----")

# 4. index() -- find the position of a specific drug
position = inventory.index("Metformin")
print(f"index('Metformin'): found at position {position}")

# 5. count() -- check how many times a drug appears (duplicate entry check)
inventory.append("Aspirin")   # simulate a duplicate entry by mistake
duplicate_count = inventory.count("Aspirin")
print(f"count('Aspirin'): appears {duplicate_count} times")

print("\n---- Dispensing and removing stock ----")

# 6. remove() -- remove a specific drug by name (e.g. recalled medication)
inventory.remove("Aspirin")   # removes only the FIRST occurrence
print(f"After remove('Aspirin') [removes first match]: {inventory}")

# 7. pop() -- remove and return an item (e.g. dispensing the last-added drug)
dispensed_drug = inventory.pop()
print(f"After pop() -> dispensed: '{dispensed_drug}', remaining: {inventory}")

# pop() with an index -- dispense a drug from a specific position
dispensed_specific = inventory.pop(1)
print(f"After pop(1) -> dispensed: '{dispensed_specific}', remaining: {inventory}")

print("\n---- Organizing the shelf ----")

# 8. sort() -- arrange drugs alphabetically for easy shelf browsing
inventory.sort()
print(f"After sort(): {inventory}")

# 9. reverse() -- reverse the current order (e.g. show newest-sorted-last first)
inventory.reverse()
print(f"After reverse(): {inventory}")

print("\n---- Backup and audit ----")

# 10. copy() -- create a backup snapshot of current inventory before an audit
inventory_backup = inventory.copy()
print(f"Backup created (copy()): {inventory_backup}")

# Prove it's an independent copy -- modifying original doesn't affect backup
inventory.append("Ranitidine")
print(f"Live inventory after new addition: {inventory}")
print(f"Backup remains unchanged: {inventory_backup}")

print("\n---- End of day: clearing a recalled batch list ----")

recalled_batch_list = ["BATCH-2026-0417", "BATCH-2026-0418"]
print(f"Recalled batch list before clear(): {recalled_batch_list}")

# 11. clear() -- empty out the list completely (e.g. recall list processed and closed)
recalled_batch_list.clear()
print(f"After clear(): {recalled_batch_list}")

print("\n---- FINAL LIVE INVENTORY ----")
print(inventory)
```

# Patient Vitals Monitoring System

```python
# ============================================
# Patient Vitals Monitoring System
# Demonstrates tuple methods & key operations
# let's call this patient_vitals_monitor_tuple.ipynb
# ============================================

# A single vitals reading -- fixed, shouldn't change once recorded (GxP audit record)
# (timestamp, systolic_bp, diastolic_bp, pulse, temp_f)
vitals_reading = ("2025-07-11T09:00", 102, 96, 78, 98.9)

print(f"Vitals reading: {vitals_reading}")

print("\n #### STEP 1: The two TRUE tuple methods ####")

# 1. count() -- how many times does a specific value appear?
readings_today = (117,80,118, 76, 72, 98.6, 72, 120, 72,120,79)   # pulse recorded 5x during the day
pulse_count = readings_today.count(72)
print(f"count(72): pulse value 72 appears {pulse_count} times today")

# 2. index() -- find the position of a specific value
position = vitals_reading.index(96)
print(f"index(96): diastolic BP found at position {position}")

print("\n----#### STEP 2: Reading data (indexing & slicing) ----####")

# 3. len() -- how many data points in this reading?
print(f"len(): {len(vitals_reading)} data points recorded")

# 4. Indexing -- access a single item by position
timestamp = vitals_reading[0]
pulse = vitals_reading[3]
print(f"Indexing: timestamp={timestamp}, pulse={pulse}")

# 5. Slicing -- get a sub-tuple (just the BP + pulse readings)
bp_and_pulse = vitals_reading[1:4]
print(f"Slicing [1:4]: {bp_and_pulse}")

# 6. Reverse slicing -- reverse the whole reading
print(f"Reversed [::-1]: {vitals_reading[::-1]}")

print("\n####---- STEP 3: Membership & combining readings ####----")

# 7. in / not in -- check if a value exists in the reading
print(f"98.6 in vitals_reading: {98.6 in vitals_reading}")
print(f"103.0 in vitals_reading: {103.0 in vitals_reading}")   # fever check

# 8. + -- concatenate two tuples (append a status flag) into a NEW tuple
vitals_with_status = vitals_reading + ("Normal",)
print(f"Concatenation (+): {vitals_with_status}")

# 9. * -- repeat a tuple (simulate 3 identical baseline placeholder readings)
baseline_placeholder = (0,) * 3
print(f"Repetition (*): {baseline_placeholder}")

print("\n#### STEP 4: Analyzing a batch of numeric readings ####")

hourly_pulse_readings = (72, 75, 71, 80, 68, 74)

# 10. sorted() -- returns a NEW sorted LIST (not a tuple!) from tuple data
print(f"sorted(): {sorted(hourly_pulse_readings)}")

# 11. sum() -- total of all numeric values
print(f"sum(): {sum(hourly_pulse_readings)}")

# 12. min() / max() -- lowest and highest pulse readings of the day
print(f"min(): {min(hourly_pulse_readings)}, max(): {max(hourly_pulse_readings)}")

print("\n----#### STEP 5: Converting & unpacking ####----")

# 13. tuple() constructor -- convert a list of readings into an immutable tuple
raw_list = [72, 75, 71]
locked_readings = tuple(raw_list)
print(f"tuple(): {locked_readings}  (type: {type(locked_readings).__name__})")

# 14. Unpacking -- assign each value to its own labeled variable
ts, sys_bp, dia_bp, pulse_val, temp = vitals_reading
print(f"Unpacked -> BP: {sys_bp}/{dia_bp}, Pulse: {pulse_val}, Temp: {temp}°F")

print("\n####STEP 6: Why tuples work as dictionary keys ####")

# 15. hash() -- tuples are hashable (lists are NOT), so they can be dict keys
print(f"hash(vitals_reading): {hash(vitals_reading)}")

# Practical use: composite key lookup for a lab reference range table
reference_ranges = {
    ("Adult", "Systolic BP"): (90, 120),
    ("Adult", "Pulse"): (60, 100),
}
print(f"Reference range lookup: {reference_ranges[('Adult', 'Pulse')]}")

print("\n#### STEP 7: Looping through readings ####")

# 16. for loop -- iterate through each value in order
labels = ("Timestamp", "Systolic", "Diastolic", "Pulse", "Temp")
for label, value in zip(labels, vitals_reading):
    print(f"  {label}: {value}")

# 17. enumerate() -- loop with position numbers, useful for a numbered report
print("\nNumbered vitals report:")
for i, value in enumerate(vitals_reading):
    print(f"  Reading #####{i+1}: {value}")
```

# Pharmacy Patient Medication Record System

```python
# ============================================
# Pharmacy Patient Medication Record System
# Demonstrates all key Python dictionary methods
# let's call this patient_med_records_dictionary
# ============================================

# A patient's medication record -- drug name mapped to dosage info
patient_meds = {
    "Metformin": "500mg twice daily",
    "Lisinopril": "10mg once daily",
    "Atorvastatin": "20mg once daily",
}
print(f"Starting record: {patient_meds}")

print("\n##### STEP 1: Safe lookups #####")

# 1. get() -- safely check a drug without crashing if it's not there
print(f"get('Metformin'): {patient_meds.get('Metformin')}")
print(f"get('Amoxicillin'): {patient_meds.get('Amoxicillin')}")               # None (no error)
print(f"get('Amoxicillin', 'Not prescribed'): {patient_meds.get('Amoxicillin', 'Not prescribed')}")

# 12. dict[key] -- direct access (raises KeyError if missing -- use carefully)
print(f"dict[key]: patient_meds['Lisinopril'] = {patient_meds['Lisinopril']}")

print("\n##### STEP 2: Viewing all keys, values, and pairs #####")

# 2. keys() -- all drug names
print(f"keys(): {(patient_meds.keys())}")

# 3. values() -- all dosage instructions
print(f"values(): {(patient_meds.values())}")

# 4. items() -- all drug-dosage pairs together
print(f"items(): {(patient_meds.items())}")

print("\n##### STEP 3: New prescription added / refill updated #####")

# 13. dict[key] = value -- add a brand-new prescription
patient_meds["Omeprazole"] = "20mg once daily"
print(f"After adding Omeprazole: {patient_meds}")

# 5. update() -- doctor changes multiple prescriptions at once
patient_meds.update({"Metformin": "850mg twice daily", "Aspirin": "81mg once daily"})
print(f"After update() [dose change + new drug]: {patient_meds}")

print("\n##### STEP 4: Discontinuing medications #####")

# 6. pop() -- discontinue a specific drug, capture what was removed for the record
discontinued = patient_meds.pop("Aspirin", "Was not prescribed")
print(f"pop('Aspirin'): discontinued '{discontinued}'")
print(f"Record now: {patient_meds}")

# 7. popitem() -- remove the most recently added entry (e.g. undo last action)
last_entry = patient_meds.popitem()
print(f"popitem(): removed most recent entry -> {last_entry}")
print(f"Record now: {patient_meds}")

# 14. del -- remove a specific entry directly
patient_meds["Ibuprofen"] = "As needed for pain"   # add one to demonstrate del
del patient_meds["Ibuprofen"]
print(f"After del: {patient_meds}")

print("\n##### STEP 5: Setting defaults for missing data #####")

# 8. setdefault() -- if a drug isn't in the record yet, add it with a default note
patient_meds.setdefault("Vitamin D", "Not yet dosed -- pending physician review")
print(f"After setdefault('Vitamin D', ...): {patient_meds}")

# calling it again on an EXISTING key does NOT overwrite it
patient_meds.setdefault("Metformin", "SHOULD NOT OVERWRITE")
print(f"setdefault on existing key (unchanged): {patient_meds['Metformin']}")

print("\n##### STEP 6: Checking membership#####")

# 15. in / 16. not in -- verify a drug is (or isn't) on the patient's record
print(f"'Metformin' in patient_meds: {'Metformin' in patient_meds}")
print(f"'Penicillin' not in patient_meds: {'Penicillin' not in patient_meds}")

print("\n##### STEP 7: Record length #####")

# 17. len() -- how many active medications does this patient have?
print(f"len(): patient is currently on {len(patient_meds)} medications")

print("\n##### STEP 8: Backing up the record before an audit #####")

# 10. copy() -- create an independent snapshot before making further changes
backup_record = patient_meds.copy()
patient_meds["Warfarin"] = "5mg once daily"   # live record changes after backup
print(f"Live record: {patient_meds}")
print(f"Backup remains unchanged: {backup_record}")

print("\n##### STEP 9: Building a new patient record from a template #####")

# 11. fromkeys() -- initialize a fresh record for a new patient with default status
new_patient_drugs = ["Metformin", "Lisinopril"]
new_patient_record = dict.fromkeys(new_patient_drugs, "Pending physician confirmation")
print(f"fromkeys(): {new_patient_record}")

print("\n##### STEP 10: Building a dict from pairs #####")

# 18. dict(iterable) -- construct a dictionary directly from a list of tuples
lab_results = dict([("Glucose", "110 mg/dL"), ("A1C", "6.2%")])
print(f"dict(iterable): {lab_results}")

print("\n#####STEP 11: Looping through the record #####")

# 19. for key in dict -- loop through drug names only
print("Drug names on file:")
for drug in patient_meds:
    print(f"  - {drug}")

# 20. for k, v in dict.items() -- loop through drug + dosage together
print("\nFull medication list:")
for drug, dosage in patient_meds.items():
    print(f"  {drug}: {dosage}")

print("\n##### STEP 12: Dictionary comprehension #####")

# 21. dict comprehension -- build a new dict, e.g. flag high-risk drugs (Warfarin, Insulin)
high_risk_drugs = {"Warfarin", "Insulin", "Digoxin"}
risk_flags = {drug: ("HIGH ALERT" if drug in high_risk_drugs else "Routine")
              for drug in patient_meds}
print(f"Risk flags: {risk_flags}")

print("\n##### STEP 13: Merging records (Python 3.9+) #####")

# 22. | -- merge two dicts into a NEW one (e.g. combine home meds + hospital meds)
hospital_meds = {"Heparin": "5000 units subcutaneous"}
combined_record = patient_meds | hospital_meds
print(f"Merged with | : {combined_record}")

# 23. |= -- merge in place, updating the original record
patient_meds |= hospital_meds
print(f"After |= (in-place merge): {patient_meds}")

print("\n##### STEP 14: End of visit -- clearing a temp record #####")

temp_visit_notes = {"Chief Complaint": "Follow-up", "BP": "118/76"}
print(f"Temp notes before clear(): {temp_visit_notes}")

# 9. clear() -- wipe out a temporary record once it's been filed elsewhere
temp_visit_notes.clear()
print(f"After clear(): {temp_visit_notes}")

print("\n##### FINAL PATIENT MEDICATION RECORD #####")
print(patient_meds)
```

# Pharmacy Drug Interaction & Comparison System

```python
# ============================================
# Pharmacy Drug Interaction & Comparison System
# let's call this pharm_drug_int_compare_set.ipynb
# Demonstrates key Python set methods
# ============================================
# This program mirrors  a realistic clinical/pharmacy workflow:

#Prescribing new drugs → add(), update() — with automatic duplicate protection (a set's core benefit)
#Discontinuing drugs → remove() vs. discard() (the classic safe-vs-risky removal distinction)
#QA sampling → pop()
#Audit backup → copy()
#Cross-patient comparison → union(), intersection(), difference(), symmetric_difference() -  how you would programmatically
#check for shared drugs or drug-interaction risk across patients
#Safety checks → issubset(), issuperset(), isdisjoint() — check if a known dangerous combo exists, or if two drug sets are
#completely unrelated
#In-place formulary filtering → |=, &=
#Fast lookups → in / not in
#Deduplicating a dispensing log → set() constructor
#Fixed interaction-pair keys → frozenset() as a dictionary key (only possible because it's immutable/hashable)
#Reporting → for loop, set comprehension

# ============================================

# Two patients' current medication sets
patient_a_drugs = {"Metformin", "Lisinopril", "Aspirin", "Ibuprofen","Keppra","pioglitazone"}
patient_b_drugs = {"Aspirin", "Warfarin", "Omeprazole","Rosuvastatin","Mucinex"}

print(f"Patient A meds: {patient_a_drugs}")
print(f"Patient B meds: {patient_b_drugs}")

print("\n##### STEP 1: Adding and updating medications #####")

# 1. add() -- prescribe a single new drug to Patient A
patient_a_drugs.add("Atorvastatin")
print(f"After add('Atorvastatin'): {patient_a_drugs}")

# 2. update() -- add multiple new prescriptions at once
patient_a_drugs.update(["Vitamin D", "Calcium"])
print(f"After update([...]): {patient_a_drugs}")

# Duplicate check -- sets automatically ignore duplicates
patient_a_drugs.add("Aspirin")   # already exists, no effect
print(f"After re-adding 'Aspirin' (no duplicate created): {patient_a_drugs}")

print("\n##### STEP 2: Discontinuing medications #####")

# 3. remove() -- discontinue a drug (raises error if not found -- use when you're SURE it exists)
patient_a_drugs.remove("Ibuprofen")
print(f"After remove('Ibuprofen'): {patient_a_drugs}")

# 4. discard() -- safely attempt to remove a drug that may or may not be present
patient_a_drugs.discard("Penicillin")   # not present -- no error, nothing happens
print(f"After discard('Penicillin') [not present, safe]: {patient_a_drugs}")

# 5. pop() -- remove an arbitrary drug (e.g. random QA sample check)
sampled_drug = patient_a_drugs.pop()
print(f"pop() removed for QA sample: '{sampled_drug}'")
print(f"Remaining: {patient_a_drugs}")

print("\n----##### STEP 3: Backup before comparison analysis ----#####")

# 7. copy() -- snapshot Patient A's record before running comparisons
patient_a_backup = patient_a_drugs.copy()
print(f"Backup created: {patient_a_backup}")

print("\n#####---- STEP 4: Comparing two patients' medications #####----")

# 8. union() / | -- combine all unique drugs across both patients
all_drugs_combined = patient_a_drugs.union(patient_b_drugs)
print(f"union() -- all unique drugs between both patients: {all_drugs_combined}")

# 9. intersection() / & -- find drugs BOTH patients are taking (useful for interaction checks)
shared_drugs = patient_a_drugs.intersection(patient_b_drugs)
print(f"intersection() -- drugs both patients share: {shared_drugs}")

# 10. difference() / - -- drugs ONLY Patient A takes (not Patient B)
only_a = patient_a_drugs.difference(patient_b_drugs)
print(f"difference() -- only Patient A takes: {only_a}")

# 11. symmetric_difference() / ^ -- drugs taken by exactly ONE patient, not both
unique_to_each = patient_a_drugs.symmetric_difference(patient_b_drugs)
print(f"symmetric_difference() -- unique to each patient: {unique_to_each}")

print("\n##### STEP 5: Checking a known drug interaction list #####")

# Known dangerous combination: Aspirin + Warfarin increases bleeding risk
high_risk_combo = {"Aspirin", "Warfarin"}

# 16. issubset() -- check if BOTH risky drugs are present in a patient's combined record
combined_meds = patient_a_drugs.union(patient_b_drugs)
has_interaction_risk = high_risk_combo.issubset(combined_meds)
print(f"issubset() -- high-risk combo present across patients?: {has_interaction_risk}")

# 17. issuperset() -- verify the full formulary covers everything a patient takes
formulary = {"Metformin", "Lisinopril", "Aspirin", "Warfarin", "Omeprazole", "Atorvastatin", "Vitamin D", "Calcium"}
formulary_covers_patient_b = formulary.issuperset(patient_b_drugs)
print(f"issuperset() -- formulary covers all of Patient B's meds?: {formulary_covers_patient_b}")

# 18. isdisjoint() -- check if two drug sets have NO overlap at all (e.g. safe combo check)
safe_otc_drugs = {"Loratadine", "Acetaminophen"}
no_overlap = patient_a_drugs.isdisjoint(safe_otc_drugs)
print(f"isdisjoint() -- no overlap with safe OTC list?: {no_overlap}")

print("\n----#####---- STEP 6: In-place updates using operators ----#####----")

new_prescriptions = {"Amlodipine"}

# 12. |= -- merge new prescriptions into Patient A's record in place
patient_a_drugs |= new_prescriptions
print(f"After |= (in-place union): {patient_a_drugs}")

# 13. &= -- keep only drugs that are ALSO on the approved insurance formulary
insurance_approved = {"Metformin", "Lisinopril", "Aspirin", "Atorvastatin", "Amlodipine", "Vitamin D"}
patient_a_drugs &= insurance_approved
print(f"After &= (in-place intersection with formulary): {patient_a_drugs}")

print("\n##### STEP 7: Membership checks #####")

# 19. in / not in -- fast lookup, e.g. allergy screening
print(f"'Aspirin' in patient_a_drugs: {'Aspirin' in patient_a_drugs}")
print(f"'Penicillin' not in patient_a_drugs: {'Penicillin' not in patient_a_drugs}")
print("\n---- STEP 8: Record size ----")

# 20. len() -- how many active medications does Patient A currently have?
print(f"len(): Patient A is on {len(patient_a_drugs)} medications")

print("\n---- STEP 9: Building sets from other data ----")

# 21. set() constructor -- deduplicate a messy dispensing log into unique drugs dispensed
dispensing_log = ["Aspirin", "Metformin", "Aspirin", "Lisinopril", "Metformin"]
unique_drugs_dispensed = set(dispensing_log)
print(f"set() from list [removes duplicates]: {unique_drugs_dispensed}")

# 22. frozenset() -- an immutable set, usable as a dictionary key (e.g. tagging a fixed interaction pair)
interaction_pair = frozenset({"Aspirin", "Warfarin"})
interaction_notes = {
    interaction_pair: "Increased bleeding risk -- monitor INR closely"
}
print(f"frozenset() as dict key: {interaction_notes[interaction_pair]}")

print("\n---- STEP 10: Looping and set comprehension ----")

# 23. for loop -- iterate through Patient A's current meds
print("Patient A's current medications:")
for drug in patient_a_drugs:
    print(f"  - {drug}")

# 24. set comprehension -- build a new set of only "high alert" drugs from the combined pool
high_alert_drugs = {"Warfarin", "Insulin", "Digoxin"}
flagged = {drug for drug in combined_meds if drug in high_alert_drugs}
print(f"\nSet comprehension -- high-alert drugs found in combined records: {flagged}")

print("\n----#####----- FINAL PATIENT A RECORD (after formulary filtering) ----#####-----")
print(patient_a_drugs)
```

**default & optional parameters based function**

```python
#####################################################################
# Default parameters — used when the caller does not provide a value
#####################################################################

# Syntax: def function(param=default_value)

def calculate_creatinine_clearance(age, weight, serum_cr, gender='male'):
    """
    Cockcroft-Gault formula for Creatinine Clearance
    Parameters:
        age       : Patient age in years (required)
        weight    : Body weight in kg (required)
        serum_cr  : Serum creatinine in mg/dL (required)
        gender    : 'male' or 'female' (optional, default='male')
    Returns:
        Creatinine clearance in mL/min
    """
    crcl = ((140 - age) * weight) / (72 * serum_cr)

    if gender == 'female':
        crcl *= 0.85   # Multiply by 0.85 for female patients

    return round(crcl, 1)

# Call without gender (uses default 'male'):
male_crcl = calculate_creatinine_clearance(55, 70, 1.2)
print(f'Male CrCl: {male_crcl} mL/min')   # Male CrCl: 45.1 mL/min

# Call with gender specified:
female_crcl = calculate_creatinine_clearance(55, 70, 1.2, gender='female')
print(f'Female CrCl: {female_crcl} mL/min') # Female CrCl: 38.3 mL/min

# Using keyword arguments — order does not matter:
result = calculate_creatinine_clearance(weight=65, age=60, serum_cr=1.5, gender='female')
print(f'CrCl: {result} mL/min')

```

# Complete Dosage Calculator based on BMI

```python
# ============================================================
# Pharmaceutical Dosage Calculator
# Uses real dosage guidelines for common drugs
# ============================================================

def calculate_bmi(weight_kg, height_m):
    """
    Calculate Body Mass Index (BMI)
    BMI = weight(kg) / height(m)^2
    """
    bmi = weight_kg / (height_m ** 2)
    return round(bmi, 1)

def classify_bmi(bmi):
    """Classify BMI into WHO categories"""
    if bmi < 18.5:
        return 'Underweight'
    elif bmi < 25.0:
        return 'Normal weight'
    elif bmi < 30.0:
        return 'Overweight'
    else:
        return 'Obese'

def get_metformin_dose(bmi, kidney_function='normal'):
    """
    Suggest Metformin starting dose based on BMI and kidney function
    Parameters:
        bmi             : Patient BMI value
        kidney_function : 'normal', 'mild', or 'moderate' (default: 'normal')
    """
    if kidney_function == 'moderate':
        return 'Metformin is CONTRAINDICATED in moderate/severe kidney impairment'

    if bmi >= 30:
        dose = '1000 mg twice daily (can escalate to 2550 mg/day)'
    elif bmi >= 25:
        dose = '500 mg twice daily or 850 mg once daily'
    else:
        dose = '500 mg once daily (lower dose for lower BMI)'

    if kidney_function == 'mild':
        dose += ' — Monitor renal function closely'

    return dose

# Main program:
print('=' * 50)
print('  PATIENT DOSAGE ASSESSMENT TOOL')
print('=' * 50)

name   = input('Patient Name: ')
weight = float(input('Weight (kg): '))
height = float(input('Height (cm): ')) / 100  # Convert cm to m
kidney = input('Kidney function (normal/mild/moderate): ')

bmi         = calculate_bmi(weight, height)
category    = classify_bmi(bmi)
dose_advice = get_metformin_dose(bmi, kidney)

print(f'\n--- Report for {name} ---')
print(f'BMI: {bmi} ({category})')
print(f'Metformin Recommendation: {dose_advice}')
```

Student Work example #2

Given a patient's current medication and a new drug being prescribed, check whether the combination appears in a known "dangerous interaction" list, and print a warning if it does.

```python
################################################################################
# Student Work Example #2: Drug Interaction Checker
################################################################################

# Define a list of known dangerous drug interaction combinations as frozensets.
# Using frozenset makes these sets immutable and hashable, suitable for use in
# other data structures if needed, and emphasizes that these combinations are fixed.
# We will use lowercase for drug names to ensure case-insensitive matching.
dangerous_interactions = [
    frozenset({"aspirin", "warfarin"}),  # Increased bleeding risk
    frozenset({"metformin", "iodinated contrast"}),  # Risk of lactic acidosis
    frozenset({"grapefruit juice", "atorvastatin"}),  # Increased statin levels
    frozenset({"ssri", "maoi"}),  # Serotonin syndrome
]

print("\n--- Drug Interaction Checker ---")

# --- Patient Data (Example) ---
# Patient's current medications (as a set for efficient lookup)
patient_current_meds = {"metformin", "amlodipine", "lisinopril", "aspirin"}
# New drug to be prescribed
new_drug = "warfarin"

# Convert new_drug to lowercase for consistent comparison
new_drug_lower = new_drug.lower()

print(f"Patient's current medications: {', '.join(patient_current_meds).title()}")
print(f"New drug to be prescribed: {new_drug.title()}")

# Combine current and new medications into a single set
# Using .union() or '|' operator on sets is efficient for combining unique elements.
all_patient_meds = patient_current_meds.union({new_drug_lower})

print(f"Combined medications (current + new): {', '.join(all_patient_meds).title()}")

# Check for dangerous interactions
interaction_found = False
for interaction_combo in dangerous_interactions:
    # If all drugs in a dangerous combination are present in the patient's combined meds
    if interaction_combo.issubset(all_patient_meds):
        # Convert the frozenset interaction to a readable string for the warning
        interacting_drugs = " and ".join([d.title() for d in interaction_combo])
        print(f"\n!!! WARNING: POTENTIAL DANGEROUS INTERACTION !!!")
        print(f"  The combination of {interacting_drugs} can be dangerous.")
        print(f"  Please consult with a physician or pharmacist before proceeding.")
        interaction_found = True
        break # Stop after finding the first dangerous interaction

if not interaction_found:
    print("\nNo dangerous interactions found with the current prescription.")
    print("Always exercise clinical judgment.")
```

```python
def calculate_dosage_times(start_hour, interval_hours, num_days):
    """
    Calculates and prints dosage times, wrapping around a 24-hour clock.

    Args:
        start_hour (int): The starting hour for the first dose (0-23).
        interval_hours (int): The interval between doses in hours.
        num_days (int): The total number of days for medication.
    """
    print(f"Dosage Schedule (starting at {start_hour}:00 and taking a dose every {interval_hours} hours for {num_days} days):")

    total_doses = (num_days * 24) // interval_hours
    current_hour = start_hour

    for i in range(total_doses):
        day_num = (current_hour // 24) + 1 # Calculate the day number
        display_hour = current_hour % 24

        # Format time to AM/PM for better readability, and handle 00:00 as 12 AM, 12:00 as 12 PM
        ampm = "AM"
        if display_hour >= 12:
            ampm = "PM"

        hour_for_display = display_hour % 12
        if hour_for_display == 0:
            hour_for_display = 12 # 00:00 becomes 12 AM, 12:00 becomes 12 PM

        print(f"Day {day_num}, Dose {i+1}: {hour_for_display:02d} {ampm} ({(current_hour % 24):02d}:00 in 24-hour format)")
        current_hour += interval_hours

# --- Example Usage ---
# Patient needs to take a tablet every 8 hours for 5 days, starting from 8 AM.
calculate_dosage_times(start_hour=8, interval_hours=8, num_days=5)
```

Student Question #4 - low stock Alert System

Given a list of drugs and their current stock counts, print an alert for any drug whose stock falls below a reorder threshold (e.g., 20 units).

```python
def check_low_stock(drug_stocks, reorder_threshold):
    """
    Checks a list of drugs and their stock counts, printing alerts
    for any drug below a specified reorder threshold.

    Args:
        drug_stocks (list of dict): A list where each dict contains
                                    'drug_name' (str) and 'stock_count' (int).
        reorder_threshold (int): The stock level below which an alert is triggered.
    """
    print(f"\n--- Low Stock Alert System (Threshold: {reorder_threshold} units) ---")
    low_stock_found = False

    for drug in drug_stocks:
        drug_name = drug['drug_name']
        stock_count = drug['stock_count']

        if stock_count < reorder_threshold:
            print(f"!!! ALERT: Low stock for {drug_name}! Current stock: {stock_count} units. Please reorder.")
            low_stock_found = True

    if not low_stock_found:
        print("All drugs are currently above the reorder threshold. No alerts.")

# --- Example Usage ---
drug_inventory = [
    {'drug_name': 'Amoxicillin', 'stock_count': 50},
    {'drug_name': 'Lisinopril', 'stock_count': 15},
    {'drug_name': 'Metformin', 'stock_count': 120},
    {'drug_name': 'Warfarin', 'stock_count': 8},
    {'drug_name': 'Ibuprofen', 'stock_count': 30},
    {'drug_name': 'Aspirin', 'stock_count': 19}
]

reorder_point = 20

check_low_stock(drug_inventory, reorder_point)
```

Student excercise #8 - Insurance Copay Calculator

Given a drug's total cost and a patient's insurance coverage percentage, calculate what the patient owes out of pocket. If the drug is on a "generic" list, apply an extra 10% discount.

```python
def calculate_copay(total_drug_cost: float, insurance_coverage_percentage: float, is_generic: bool) -> float:
    """
    Calculates the patient's out-of-pocket cost for a drug, considering insurance
    coverage and an additional discount for generic drugs.

    Args:
        total_drug_cost (float): The total cost of the drug.
        insurance_coverage_percentage (float): The percentage of the cost covered by insurance (e.g., 0.8 for 80%).
        is_generic (bool): True if the drug is generic, False otherwise.

    Returns:
        float: The final amount the patient owes out of pocket.
    """
    # Calculate the amount covered by insurance
    covered_amount = total_drug_cost * insurance_coverage_percentage

    # Calculate the initial out-of-pocket amount
    patient_owes = total_drug_cost - covered_amount

    # Apply an extra 10% discount if the drug is generic
    if is_generic:
        generic_discount = patient_owes * 0.10
        patient_owes -= generic_discount

    return round(patient_owes, 2)

# --- Example Usage ---
print("--- Insurance Copay Calculator ---")

# Scenario 1: Non-generic drug, 70% coverage
drug_cost_1 = 150.00
coverage_1 = 0.70
is_generic_1 = False
copay_1 = calculate_copay(drug_cost_1, coverage_1, is_generic_1)
print(f"Scenario 1 (Non-generic, 70% coverage):\n  Drug Cost: ${drug_cost_1:.2f}\n  Coverage: {coverage_1*100:.0f}%\n  Generic: {is_generic_1}\n  Patient Owes: ${copay_1:.2f}\n")

# Scenario 2: Generic drug, 80% coverage
drug_cost_2 = 50.00
coverage_2 = 0.80
is_generic_2 = True
copay_2 = calculate_copay(drug_cost_2, coverage_2, is_generic_2)
print(f"Scenario 2 (Generic, 80% coverage):\n  Drug Cost: ${drug_cost_2:.2f}\n  Coverage: {coverage_2*100:.0f}%\n  Generic: {is_generic_2}\n  Patient Owes: ${copay_2:.2f}\n")

# Scenario 3: Non-generic drug, 100% coverage (should be $0 out of pocket)
drug_cost_3 = 200.00
coverage_3 = 1.00
is_generic_3 = False
copay_3 = calculate_copay(drug_cost_3, coverage_3, is_generic_3)
print(f"Scenario 3 (Non-generic, 100% coverage):\n  Drug Cost: ${drug_cost_3:.2f}\n  Coverage: {coverage_3*100:.0f}%\n  Generic: {is_generic_3}\n  Patient Owes: ${copay_3:.2f}\n")

# Scenario 4: Generic drug, 50% coverage
drug_cost_4 = 100.00
coverage_4 = 0.50
is_generic_4 = True
copay_4 = calculate_copay(drug_cost_4, coverage_4, is_generic_4)
print(f"Scenario 4 (Generic, 50% coverage):\n  Drug Cost: ${drug_cost_4:.2f}\n  Coverage: {coverage_4*100:.0f}%\n  Generic: {is_generic_4}\n  Patient Owes: ${copay_4:.2f}\n")

#User input based calculation scenario

# Scenario 5: User input driven dynamic calculatons
drug_cost_input = input("Enter the drug cost: ")
drug_cost_input = float(drug_cost_input)
coverage_input = input("Enter the coverage percentage: ")
coverage_input = float(coverage_input) / 100
is_generic_flag = input("Is the drug generic? (yes/no): ")
is_generic_flag = is_generic_flag.lower() == 'yes'
copay_4 = calculate_copay(drug_cost_input, coverage_input, is_generic_flag)
print(f"Scenario 4 (Generic, 50% coverage):\n  Drug Cost: ${drug_cost_4:.2f}\n  Coverage: {coverage_4*100:.0f}%\n  Generic: {is_generic_4}\n  Patient Owes: ${copay_4:.2f}\n")
```

**Student Question #9 Adverse event key word scanner**

Given a patient's free-text symptom note (a string), scan it for known adverse-event keywords (e.g., "rash", "nausea", "dizziness") and print which ones were found - a simplified pharmacovigilance tool.

```python
def scan_adverse_events(symptom_note: str) -> None:
    """
    Scans a patient's free-text symptom note for known adverse-event keywords
    and prints which ones were found.

    Args:
        symptom_note (str): The free-text symptom description from the patient.
    """
    # Define a list of common adverse event keywords (case-insensitive)
    adverse_event_keywords = [
        "rash", "nausea", "dizziness", "vomiting", "diarrhea",
        "headache", "fatigue", "fever", "anaphylaxis", "swelling",
        "itching", "difficulty breathing", "hives", "muscle pain","weakness","blood loss"
    ]

    found_events = []
    note_lower = symptom_note.lower() # Convert the note to lowercase for case-insensitive matching

    print(f"\n--- Scanning Symptom Note: '{symptom_note}' ---")

    for keyword in adverse_event_keywords:
        if keyword in note_lower:
            found_events.append(keyword)

    if found_events:
        print("Potential Adverse Events Detected:")
        for event in found_events:
            print(f"- {event.capitalize()}")
    else:
        print("No known adverse events detected.")

# --- Example Usage ---
# Scenario 1: Multiple adverse events
patient_note_1 = "Patient reports severe headache and persistent nausea after medication."
scan_adverse_events(patient_note_1)

# Scenario 2: Single adverse event
patient_note_2 = "Developed a mild rash on the arm this morning."
scan_adverse_events(patient_note_2)

# Scenario 3: No adverse events
patient_note_3 = "Patient feels much better, no new symptoms."
scan_adverse_events(patient_note_3)

# Scenario 4: Adverse event with different casing
patient_note_4 = "Complaint of Dizziness and some faTigue throughout the day."
scan_adverse_events(patient_note_4)

# Scenario 5: More complex note with various symptoms
patient_note_5 = "Initial improvement, but then developed a new swelling in the ankles and some diarrhea."
scan_adverse_events(patient_note_5)
```

# Hospital Network Problem scenario

Problem Statement: A hospital network has two pharmacies that each maintain their own list of patient-reported allergies for a shared patient (since the patient has visited both locations). Write a program that takes both allergy lists and identifies: (a) allergies reported at both pharmacies (confirmed, high-confidence allergies), and (b) allergies reported at only one location (which may need follow-up verification before the patient's next prescription is filled).

```python
################################################################
#A hospital network has two pharmacies that each maintain their own list of patient-reported
#allergies for a shared patient (since the patient has visited both locations).
#Write a program that takes both allergy lists and identifies:
#(a) allergies reported at both pharmacies (confirmed, high-confidence allergies), and
#(b) allergies reported at only one location (which may need follow-up verification
# before the patient's next prescription is filled).
################################################################
#let's define a function called check_allergy_overlap that takes 2 parameters
def check_allergy_overlap(pharmacy_a, pharmacy_b):
    """
    Compares allergy records from two pharmacies for a shared patient.

    Parameters:
        pharmacy_a (set): allergies reported at Pharmacy A
        pharmacy_b (set): allergies reported at Pharmacy B

    Returns:
        dict: confirmed (both), needs_verification (only one), master_list (all, deduplicated)
    """
    confirmed = pharmacy_a & pharmacy_b          # intersection: reported at BOTH locations
    needs_verification = pharmacy_a ^ pharmacy_b  # symmetric difference: reported at only ONE location
    master_list = pharmacy_a | pharmacy_b         # union: complete deduplicated allergy list

    return {
        "confirmed": confirmed,
        "needs_verification": needs_verification,
        "master_list": master_list,
    }

# Example data — allergy records for the same patient at two pharmacy locations
pharmacy_a = {"Penicillin", "Sulfa", "Latex"}
pharmacy_b = {"Penicillin", "Aspirin"}

result = check_allergy_overlap(pharmacy_a, pharmacy_b)

print("Pharmacy A reported allergies:", pharmacy_a)
print("Pharmacy B reported allergies:", pharmacy_b)
print()
print("✅ Confirmed allergies (reported at both pharmacies):")
print(" ", result["confirmed"])

print()
print("⚠️  Needs verification (Allergies reported at only one pharmacy):")
print(" ", result["needs_verification"])

print()
print("📋 Unified master allergy list (for patient's combined record):")
print(" ", result["master_list"])
```

```python
import numpy as np
import urllib.request
import os

# -----------------------------------------------------------------------
# STEP 1: Download the dataset (only if we don't already have it)
# -----------------------------------------------------------------------
# The dataset is a plain CSV file with NO header row. Each line is one
# patient, with 9 comma-separated numbers.
DATA_URL = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/pima-indians-diabetes.data.csv"
LOCAL_FILE = "diabetes.csv"

if not os.path.exists(LOCAL_FILE):
    print("Downloading dataset...")
    urllib.request.urlretrieve(DATA_URL, LOCAL_FILE)
    print("Download complete.\n")
else:
    print(f"Using already-downloaded file: {LOCAL_FILE}\n")

# -----------------------------------------------------------------------
# STEP 2: Load the CSV into a NumPy array
# -----------------------------------------------------------------------
# np.loadtxt() reads a text file straight into a 2D NumPy array.
# delimiter="," tells it the values are comma-separated.
data = np.loadtxt(LOCAL_FILE, delimiter=",")

print("STEP 2: Data loaded into a NumPy array")
print("-" * 60)
print(f"Shape of the array (rows, columns): {data.shape}")
print("  -> 768 patients, 9 columns (8 measurements + 1 outcome)\n")

# Split into features (X) and target (y)
X = data[:, :-1]  # All rows, all columns except the last one
y = data[:, -1]   # All rows, only the last column (the outcome)

# -----------------------------------------------------------------------
# STEP 3: Give the columns meaningful names
# -----------------------------------------------------------------------
# NumPy arrays don't have column labels like a spreadsheet, so we keep
# a separate list of names in the SAME ORDER as the columns in the file.
column_names = [
    "Pregnancies",              # column 0
    "Glucose",                  # column 1  (plasma glucose concentration)
    "BloodPressure",            # column 2  (diastolic, mm Hg)
    "SkinThickness",            # column 3  (triceps skin fold, mm)
    "Insulin",                  # column 4  (2-hour serum insulin, mu U/ml)
    "BMI",                      # column 5  (body mass index)
    "DiabetesPedigreeFunction", # column 6  (genetic risk score)
    "Age",                      # column 7  (years)
    "Outcome",                  # column 8  (0 = non-diabetic, 1 = diabetic)
]

print("STEP 3: Column reference")
print("-" * 60)
for i, name in enumerate(column_names):
    print(f"  Column {i}: {name}")
print()

# -----------------------------------------------------------------------
# STEP 4: Basic descriptive statistics across the WHOLE dataset
# -----------------------------------------------------------------------
# axis=0 tells NumPy "calculate down each column" (i.e. per measurement),
# rather than across each row (per patient).
print("STEP 4: Descriptive statistics for each measurement")
print("-" * 60)
print(f"{'Measurement':<26}{'Mean':>10}{'Std Dev':>10}{'Min':>8}{'Max':>8}")
for i, name in enumerate(column_names[:-1]):   # skip Outcome column
    col = X[:, i]
    print(f"{name:<26}{np.mean(col):>10.2f}{np.std(col):>10.2f}"
          f"{np.min(col):>8.1f}{np.max(col):>8.1f}")
print()

# -----------------------------------------------------------------------
# STEP 5: Data quality check — this dataset uses 0 as a placeholder for
# MISSING values in columns where a true zero is medically impossible
# (nobody has a blood pressure or BMI of exactly 0).
# -----------------------------------------------------------------------
print("STEP 5: Data quality check — impossible zero values (likely missing data)")
print("-" * 60)
cols_to_check = ["Glucose", "BloodPressure", "SkinThickness", "Insulin", "BMI"]
for name in cols_to_check:
    idx = column_names.index(name)
    zero_count = np.sum(X[:, idx] == 0)          # np.where-style boolean count
    pct = (zero_count / X.shape[0]) * 100
    print(f"  {name:<16}: {zero_count:>4} zero entries  ({pct:5.1f}% of patients)")
print("  -> In real analysis, these would be treated as missing (NaN),")
print("     not as valid clinical readings, before running statistics.\n")

# -----------------------------------------------------------------------
# STEP 6: Compare diabetic vs non-diabetic patients using Boolean indexing
# -----------------------------------------------------------------------
# y == 1 creates a Boolean array (True/False for every patient).
# Using that Boolean array inside [ ] filters the rows that match.
diabetic = X[y == 1]        # rows where the outcome is 1 (diabetic)
non_diabetic = X[y == 0]    # rows where the outcome is 0 (non-diabetic)

print("STEP 6: Diabetic vs. Non-diabetic group comparison")
print("-" * 60)
print(f"Diabetic patients:     {diabetic.shape[0]}")
print(f"Non-diabetic patients: {non_diabetic.shape[0]}\n")

compare_cols = ["Glucose", "BMI", "Age", "Pregnancies"]
print(f"{'Measurement':<16}{'Diabetic Avg':>16}{'Non-Diabetic Avg':>20}{'Difference':>14}")
for name in compare_cols:
    idx = column_names.index(name)
    d_avg = np.mean(diabetic[:, idx])
    n_avg = np.mean(non_diabetic[:, idx])
    print(f"{name:<16}{d_avg:>16.2f}{n_avg:>20.2f}{(d_avg - n_avg):>14.2f}")
print()

# -----------------------------------------------------------------------
# STEP 7: Flag high-risk patients using np.where()
# -----------------------------------------------------------------------
# np.where(condition, value_if_true, value_if_false) works like a
# column-by-column if/else, applied to the whole array at once.
glucose = X[:, column_names.index("Glucose")]
risk_flag = np.where(glucose > 140, "HIGH", "NORMAL")   # simple clinical threshold

high_risk_count = np.sum(risk_flag == "HIGH")
print("STEP 7: Flagging high glucose readings (> 140 mg/dL)")
print("-" * 60)
print(f"Patients flagged HIGH glucose: {high_risk_count} out of {len(glucose)} "
      f"({(high_risk_count/len(glucose))*100:.1f}%)\n")

# -----------------------------------------------------------------------
# STEP 8: Correlation between Glucose and Outcome
# -----------------------------------------------------------------------
# np.corrcoef() returns a matrix; the value at [0,1] is the correlation
# between the two variables we passed in. Ranges from -1 to 1.
correlation_matrix = np.corrcoef(glucose, y)
glucose_outcome_corr = correlation_matrix[0, 1]

print("STEP 8: Correlation between Glucose level and Diabetes Outcome")
print("-" * 60)
print(f"Correlation coefficient: {glucose_outcome_corr:.3f}")
print("  -> Closer to +1 means higher glucose is strongly associated")
print("     with a positive diabetes outcome in this dataset.\n")

# -----------------------------------------------------------------------
# STEP 9: Percentiles — useful for setting clinical reference ranges
# -----------------------------------------------------------------------
bmi = X[:, column_names.index("BMI")]
p25, p50, p75, p95 = np.percentile(bmi, [25, 50, 75, 95])

print("STEP 9: BMI percentiles across all patients")
print("-" * 60)
print(f"25th percentile: {p25:.1f}")
print(f"50th percentile (median): {p50:.1f}")
print(f"75th percentile: {p75:.1f}")
print(f"95th percentile: {p95:.1f}\n")

# -----------------------------------------------------------------------
# STEP 10: Summary printout
# -----------------------------------------------------------------------
print("=" * 60)
print("SUMMARY")
print("=" * 60)
print(f"Total patients analyzed : {data.shape[0]}")
print(f"Diabetic                : {diabetic.shape[0]} "
      f"({diabetic.shape[0]/data.shape[0]*100:.1f}%)")
print(f"Non-diabetic             : {non_diabetic.shape[0]} "
      f"({non_diabetic.shape[0]/data.shape[0]*100:.1f}%)")
print(f"Avg glucose (diabetic)   : {np.mean(diabetic[:, 1]):.1f} mg/dL")
print(f"Avg glucose (non-diabetic): {np.mean(non_diabetic[:, 1]):.1f} mg/dL")
print(f"Glucose–Outcome correlation: {glucose_outcome_corr:.3f}")
print("=" * 60)


```

Filtering , Grouping & Aggregation

```python
import pandas as pd

# Sample clinical trial dataset:
data = {
    'Drug':     ['Metformin','Metformin','Aspirin','Aspirin','Metformin','Aspirin'],
    'Gender':   ['M','F','M','F','F','M'],
    'Age':      [45, 52, 67, 41, 38, 55],
    'Dose_mg':  [500, 850, 75, 75, 500, 150],
    'HbA1c_pre': [8.5, 9.0, 6.0, 6.2, 8.8, 5.9],
    'HbA1c_post': [7.2, 7.8, 5.8, 6.0, 7.5, 5.7],
    'ADR': [False, True, False, False, True, False]
}
df = pd.DataFrame(data)

# --- Filtering rows ---
# Single condition:
metformin_patients = df[df['Drug'] == 'Metformin']
print('Metformin patients:', len(metformin_patients))

# Multiple conditions (use & for AND, | for OR):
female_metformin = df[(df['Drug'] == 'Metformin') & (df['Gender'] == 'F')]
print('Female Metformin patients:',len(female_metformin))

# Patients who had an ADR (Adverse Drug Reaction):
adr_patients = df[df['ADR'] == True]
print('Patients with ADR:', len(adr_patients))

# --- Grouping and Aggregation ---
# groupby() — group data by a column, then apply a function

# Average HbA1c reduction by drug:
df['HbA1c_change'] = df['HbA1c_pre'] - df['HbA1c_post']

drug_stats = df.groupby('Drug')['HbA1c_change'].agg(['mean', 'std', 'count'])
drug_stats.columns = ['Avg_Reduction', 'Std_Dev', 'Patient_Count']
print('\n--- HbA1c Reduction by Drug ---')
print(drug_stats.round(2))

# ADR rate by drug:
adr_rate = df.groupby('Drug')['ADR'].mean() * 100  # As percentage
print('\n--- ADR Rate by Drug (%) ---')
print(adr_rate.round(1))

# Multiple aggregations:
summary = df.groupby('Drug').agg({
    'Age': 'mean',
    'Dose_mg': 'median',
    'HbA1c_change': 'mean',
    'ADR': 'sum'
})
print('\n--- Complete Drug Summary ---')
print(summary.round(2))
```

The above reorder point formula ROP = μ + (Z × σ) treats demand during lead time as a normally distributed random variable, where μ is the expected demand and σ captures its variability.

## Task 1: Data Preprocessing & Cleaning

```python
import pandas as pd
import numpy as np
import kagglehub
import os

# --- 1. Load and Inspect: Load salesdaily.csv ---

print("Downloading Kaggle dataset 'milanzdravkovic/pharma-sales-data'...")
# Download the dataset using kagglehub
path = kagglehub.dataset_download("milanzdravkovic/pharma-sales-data")

# Construct the full path to the salesdaily.csv file within the downloaded directory
csv_file_path = os.path.join(path, 'salesdaily.csv')

# Load the dataset
df = pd.read_csv(csv_file_path)
print("Dataset loaded successfully!")

print("\nDataset Shape:", df.shape)
print("\nData Types and Non-Null Counts:")
df.info()
print("\nMissing Values per Column:")
print(df.isnull().sum())

# Drop pre-existing time-related columns that will be extracted in Task 2
# These columns 'Year', 'Month', 'Hour', 'Weekday Name' are present in the raw data but we want to recreate them in Task 2
df = df.drop(columns=['Year', 'Month', 'Hour', 'Weekday Name'], errors='ignore')

# --- 2. Date Alignment: Convert date to datetime and set as index ---
# The dataset has 'datum' column, we will use 'format="mixed"' to infer the date format dynamically
df['datum'] = pd.to_datetime(df['datum'], format='mixed')
df = df.set_index('datum')
print("\nDate column converted to datetime and set as index.")
print(df.head())

# --- 3. Outlier Mitigation with NumPy ---
# The problem states 'sales figures that are exceptionally high (e.g., more than 3 standard deviations from the mean for a specific drug category)'.
# We will apply this per drug category column.

# Identify drug category columns (assuming all columns except 'datum' which is now the index and the pre-existing time features)
drug_columns = ['M01AB', 'M01AE', 'N02BA', 'N02BE', 'N05B', 'N05C', 'R03', 'R06']

for col in drug_columns:
    # Calculate mean and standard deviation for the current drug category
    mean_sales = df[col].mean()
    std_sales = df[col].std()

    # Define the upper threshold for outliers (mean + 3 * std)
    outlier_threshold = mean_sales + (3 * std_sales)

    # Calculate the median for the current drug category
    median_sales = df[col].median()

    # Use np.where to replace values above the threshold with the median
    # Only apply if there are actual outliers to avoid unnecessary computation
    if (df[col] > outlier_threshold).any():
        original_outliers = df[df[col] > outlier_threshold][col].count()
        df[col] = np.where(df[col] > outlier_threshold, median_sales, df[col])
        print(f"\nOutliers replaced in column '{col}': {original_outliers} values replaced with median {median_sales:.2f}")
    else:
        print(f"\nNo significant outliers found in column '{col}' above 3 standard deviations.")

print("\nOutlier mitigation complete. Displaying info after cleaning:")
df.info()
```

## Task 2: Temporal & Seasonal Analysis

## Task 4: Dynamic Inventory Thresholds

```python
# --- 1. Calculate mean (μ) and standard deviation (σ) of daily sales for each drug class ---
# We'll use the same existing_drug_cols from Task 3, or all numeric columns if Task 3 wasn't executed.

if 'existing_drug_cols' not in locals():
    # If Task 3 wasn't run, define drug_category_columns again
    drug_category_columns = ['M01AE', 'N02BA', 'N02BE', 'N05B', 'N05C', 'R03', 'R06']
    existing_drug_cols = [col for col in drug_category_columns if col in df.columns]

if existing_drug_cols:
    daily_sales_stats = df[existing_drug_cols].agg(['mean', 'std']).T
    daily_sales_stats.columns = ['Mean_Daily_Sales', 'Std_Dev_Daily_Sales']
    print("\nMean (μ) and Standard Deviation (σ) of daily sales per drug class:")
    print(daily_sales_stats.round(2))

    # --- 2. Calculate Reorder Point (ROP) for each drug class ---
    # ROP = μ + (1.65 * σ)
    # Z = 1.65 for a 95% service level (5% stockout probability)
    Z_score = 1.65

    daily_sales_stats['Reorder_Point'] = daily_sales_stats['Mean_Daily_Sales'] + (Z_score * daily_sales_stats['Std_Dev_Daily_Sales'])

    # --- 3. Output a summarized dictionary or DataFrame showing the calculated Reorder Point ---
    print("\nCalculated Reorder Points (ROP) for all drug classes:")
    print(daily_sales_stats[['Reorder_Point']].round(2))

    # Convert to dictionary for easy access, if preferred
    reorder_points_dict = daily_sales_stats['Reorder_Point'].round(2).to_dict()
    print("\nReorder Points as Dictionary:")
    print(reorder_points_dict)
else:
    print("\nNo drug category columns available for inventory threshold calculation.")
```

Sample code that fetches the top adverse reactions reported for a given drug from the FDA's openFDA Drug Adverse Event API and returns them as a pandas DataFrame.

```python
"""
FAERS Adverse Event Lookup (openFDA API)
------------------------------------------
Fetches the top adverse reactions reported for a given drug from the
FDA's openFDA Drug Adverse Event API and returns them as a pandas DataFrame.

Documentation: https://open.fda.gov/apis/

Enhancements over the original script:
  1. Input validation & sanitization for the drug name
  2. Explicit request timeout (prevents the script hanging forever)
  3. Retry logic with exponential backoff for transient network/server errors
  4. Granular exception handling (connection, timeout, HTTP, JSON, schema)
  5. HTTPS enforced; response size capped to avoid memory-exhaustion issues
  6. Optional API key read from an environment variable (never hardcoded)
  7. Logging instead of bare print() for real diagnostics
  8. No use of eval/exec/str-formatted SQL/etc. — all params passed via
     `requests`' own `params=` dict, which handles URL-encoding for you
     and avoids manual string concatenation into the URL (a common
     injection vector when people build query strings by hand).
"""

import logging
import os
import re
import time
from dataclasses import dataclass
from typing import Optional

import pandas as pd
import requests
from requests.adapters import HTTPAdapter
from urllib3.util.retry import Retry

# --------------------------------------------------------------------------
# Configuration
# --------------------------------------------------------------------------

FAERS_URL = "https://api.fda.gov/drug/event.json"  # always HTTPS
REQUEST_TIMEOUT_SECONDS = 10          # (connect timeout, read timeout)
MAX_RETRIES = 3
BACKOFF_FACTOR = 1.5                  # 0s, 1.5s, 3s... between retries
MAX_LIMIT = 100                       # openFDA's practical per-request cap
MAX_RESPONSE_BYTES = 5_000_000        # 5 MB safety cap on response size

# openFDA works fine without a key for light use, but an optional key
# raises your rate limit. NEVER hardcode a real key in source code —
# read it from the environment (or a secrets manager) instead.
API_KEY = os.environ.get("OPENFDA_API_KEY")  # None if not set

# Only allow letters, numbers, spaces, and a few safe punctuation marks
# in a drug name. This isn't "SQL injection" territory since `requests`
# URL-encodes params for us, but validating input still protects you
# against malformed/oversized queries and unexpected API errors.
_DRUG_NAME_PATTERN = re.compile(r"^[A-Za-z0-9 \-']{1,100}$")

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s [%(levelname)s] %(message)s",
)
logger = logging.getLogger("faers_lookup")


# --------------------------------------------------------------------------
# Custom exceptions — lets callers distinguish failure types cleanly
# --------------------------------------------------------------------------

class FAERSLookupError(Exception):
    """Base class for all errors raised by this module."""


class InvalidDrugNameError(FAERSLookupError):
    """Raised when the supplied drug name fails validation."""


class FAERSAPIError(FAERSLookupError):
    """Raised when the openFDA API returns an error or malformed response."""


@dataclass
class AdverseEventResult:
    drug_name: str
    dataframe: pd.DataFrame
    total_reports_returned: int


# --------------------------------------------------------------------------
# Core logic
# --------------------------------------------------------------------------

def _validate_drug_name(drug_name: str) -> str:
    """Validate and normalize a user-supplied drug name.

    Raises InvalidDrugNameError if the name looks unsafe or malformed.
    """
    if not isinstance(drug_name, str):
        raise InvalidDrugNameError("Drug name must be a string.")

    cleaned = drug_name.strip()

    if not cleaned:
        raise InvalidDrugNameError("Drug name cannot be empty.")

    if not _DRUG_NAME_PATTERN.match(cleaned):
        raise InvalidDrugNameError(
            f"Drug name '{drug_name}' contains unsupported characters. "
            "Only letters, numbers, spaces, hyphens, and apostrophes are allowed."
        )

    return cleaned.lower()


def _build_session() -> requests.Session:
    """Create a requests Session configured with automatic retries.

    Retries only on transient failures (connection errors, 5xx server
    errors, and 429 rate-limit responses) — never on 4xx client errors
    like a bad request, since retrying those won't help.
    """
    session = requests.Session()
    retry_strategy = Retry(
        total=MAX_RETRIES,
        backoff_factor=BACKOFF_FACTOR,
        status_forcelist=[429, 500, 502, 503, 504],
        allowed_methods=["GET"],
        raise_on_status=False,
    )
    adapter = HTTPAdapter(max_retries=retry_strategy)
    session.mount("https://", adapter)
    # Deliberately do NOT mount a plain "http://" adapter — this ensures
    # any accidental non-HTTPS URL fails fast instead of sending data
    # in the clear.
    session.headers.update({
        "User-Agent": "Pocket-Python-Pharma-Textbook/1.0 (educational use)",
        "Accept": "application/json",
    })
    return session


def fetch_adverse_events(
    drug_name: str,
    limit: int = 10,
    session: Optional[requests.Session] = None,
) -> AdverseEventResult:
    """Fetch the top reported adverse reactions for a drug from FAERS.

    Parameters
    ----------
    drug_name : str
        Name of the drug to search for (e.g. "metformin").
    limit : int
        Number of top adverse-reaction categories to return (1-100).
    session : requests.Session, optional
        Reuse an existing session (e.g. across multiple lookups) instead
        of creating a new one each call.

    Returns
    -------
    AdverseEventResult
        Structured result containing the drug name, a DataFrame of
        (Adverse_Reaction, Report_Count), and the row count returned.

    Raises
    ------
    InvalidDrugNameError
        If drug_name or limit fails validation.
    FAERSAPIError
        If the API call fails, times out, or returns malformed data.
    """
    validated_name = _validate_drug_name(drug_name)

    if not isinstance(limit, int) or not (1 <= limit <= MAX_LIMIT):
        raise InvalidDrugNameError(
            f"'limit' must be an integer between 1 and {MAX_LIMIT}."
        )

    params = {
        "search": f"patient.drug.medicinalproduct:{validated_name}",
        "count": "patient.reaction.reactionmeddrapt.exact",
        "limit": limit,
    }
    if API_KEY:
        params["api_key"] = API_KEY

    owns_session = session is None
    session = session or _build_session()

    try:
        logger.info("Requesting FAERS adverse-event data for '%s'...", validated_name)
        response = session.get(
            FAERS_URL,
            params=params,
            timeout=REQUEST_TIMEOUT_SECONDS,
        )

        # Guard against an unexpectedly huge payload before parsing it.
        content_length = response.headers.get("Content-Length")
        if content_length and int(content_length) > MAX_RESPONSE_BYTES:
            raise FAERSAPIError(
                f"Response too large ({content_length} bytes) — refusing to parse."
            )

        # openFDA returns 404 (not an error status per se) when a search
        # matches zero records — handle that distinctly from real failures.
        if response.status_code == 404:
            logger.warning("No FAERS records found for '%s'.", validated_name)
            return AdverseEventResult(
                drug_name=validated_name,
                dataframe=pd.DataFrame(columns=["Adverse_Reaction", "Report_Count"]),
                total_reports_returned=0,
            )

        response.raise_for_status()  # raises HTTPError for 4xx/5xx (except 404, handled above)

        try:
            data = response.json()
        except ValueError as exc:
            raise FAERSAPIError("openFDA response was not valid JSON.") from exc

        results = data.get("results")
        if results is None:
            raise FAERSAPIError("Unexpected API response: missing 'results' field.")

        df = pd.DataFrame(results)
        if df.empty:
            df = pd.DataFrame(columns=["Adverse_Reaction", "Report_Count"])
        else:
            # Defensive rename: only proceed if the expected columns are present,
            # rather than blindly assuming column order/count as the original did.
            expected_cols = {"term", "count"}
            if not expected_cols.issubset(df.columns):
                raise FAERSAPIError(
                    f"Unexpected result schema. Got columns: {list(df.columns)}"
                )
            df = df.rename(columns={"term": "Adverse_Reaction", "count": "Report_Count"})
            df = df.sort_values("Report_Count", ascending=False).reset_index(drop=True)

        logger.info(
            "Retrieved %d adverse-reaction categories for '%s'.",
            len(df), validated_name,
        )
        return AdverseEventResult(
            drug_name=validated_name,
            dataframe=df,
            total_reports_returned=len(df),
        )

    except requests.exceptions.Timeout as exc:
        raise FAERSAPIError(
            f"Request to openFDA timed out after {REQUEST_TIMEOUT_SECONDS}s."
        ) from exc
    except requests.exceptions.ConnectionError as exc:
        raise FAERSAPIError(
            "Could not connect to openFDA. Check your network connection."
        ) from exc
    except requests.exceptions.HTTPError as exc:
        status = exc.response.status_code if exc.response is not None else "unknown"
        raise FAERSAPIError(f"openFDA returned an HTTP error (status {status}).") from exc
    except requests.exceptions.RequestException as exc:
        # Catch-all for any other requests-library failure we didn't
        # anticipate, so callers never see a raw, unfiltered exception type.
        raise FAERSAPIError(f"Unexpected request failure: {exc}") from exc
    finally:
        if owns_session:
            session.close()


# --------------------------------------------------------------------------
# Script entry point
# --------------------------------------------------------------------------

def main() -> None:
    drug_name = "metformin"

    try:
        result = fetch_adverse_events(drug_name, limit=10)
    except InvalidDrugNameError as exc:
        logger.error("Invalid input: %s", exc)
        return
    except FAERSAPIError as exc:
        logger.error("Lookup failed: %s", exc)
        return

    if result.total_reports_returned == 0:
        print(f"No adverse event data found for '{result.drug_name}'.")
        return

    print(f"\nTop {result.total_reports_returned} Adverse Reactions for "
          f"{result.drug_name.title()} (FDA FAERS):")
    print(result.dataframe.to_string(index=False))


if __name__ == "__main__":
    main()
```

## Final Mini Project: Pima Indians Diabetes Data Analysis

### Task 1: Load the Data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import urllib.request
import os

# Source URL for the dataset
DATA_URL = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/pima-indians-diabetes.data.csv"
LOCAL_FILE = "pima-indians-diabetes.csv"

# Download the dataset if it doesn't already exist locally
if not os.path.exists(LOCAL_FILE):
    print(f"Downloading dataset from {DATA_URL}...")
    urllib.request.urlretrieve(DATA_URL, LOCAL_FILE)
    print("Download complete.")
else:
    print(f"Using already-downloaded file: {LOCAL_FILE}")


# Define column names as the dataset has no header
column_names = [
    "Pregnancies",              # Number of times pregnant
    "Glucose",                  # Plasma glucose concentration a 2 hours in an oral glucose tolerance test
    "BloodPressure",            # Diastolic blood pressure (mm Hg)
    "SkinThickness",            # Triceps skin fold thickness (mm)
    "Insulin",                  # 2-Hour serum insulin (mu U/ml)
    "BMI",                      # Body mass index (weight in kg/(height in m)^2)
    "DiabetesPedigreeFunction", # Diabetes pedigree function
    "Age",                      # Age (years)
    "Outcome"                   # Class variable (0 = non-diabetic, 1 = diabetic)
]

# Load the CSV into a pandas DataFrame
df = pd.read_csv(LOCAL_FILE, names=column_names)

# Confirm the shape of the DataFrame
print(f"\nDataset loaded. Shape: {df.shape} (rows, columns)")

# Display the first 5 rows to verify loading and column names
print("\nFirst 5 rows of the dataset:")
display(df.head())
```

### Task 2: Explore Before Touching Anything

```python
print("\nDataFrame Info (data types and non-null counts):")
df.info()

print("\nDataFrame Data Types:")
print(df.dtypes)

print("\nDescriptive Statistics for Numerical Columns:")
display(df.describe())
```

### Task 3: Audit for Data Quality Issues

```python
print("\n--- Data Quality Audit: Identifying Implausible Zero Values ---\n")

# Columns where a value of zero is not biologically plausible and likely indicates missing data
columns_with_implausible_zeros = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']

# Loop through each identified column to count zero values
for col in columns_with_implausible_zeros:
    zero_count = (df[col] == 0).sum()
    total_rows = len(df)
    percentage = (zero_count / total_rows) * 100

    print(f"Column '{col}': {zero_count} zero entries ({percentage:.2f}% of patients)")

print("\nThese zero values will be treated as missing data in the cleaning phase.")
```

### Task 5: Engineer a Risk Feature

```python
print("\n--- Engineering 'BMI_Category' Feature ---\n")

# Using np.where to categorize BMI values into 'Underweight', 'Normal weight', 'Overweight', 'Obese'
# BMI Categories (WHO):
#   Underweight: < 18.5
#   Normal weight: 18.5 – 24.9
#   Overweight: 25.0 – 29.9
#   Obese: >= 30.0

df['BMI_Category'] = np.where(df['BMI'] < 18.5, 'Underweight',
                     np.where(df['BMI'] < 25.0, 'Normal weight',
                     np.where(df['BMI'] < 30.0, 'Overweight',
                     'Obese')))

print("Created 'BMI_Category' column based on WHO classifications.")

print("\nValue counts for new 'BMI_Category' column:")
display(df['BMI_Category'].value_counts())

print("\nFirst 5 rows with new 'BMI_Category' column:")
display(df.head())
```

### Task 4: Clean the Dataset

```python
print("\n--- Cleaning the Dataset: Handling Implausible Zero Values ---\n")

# Columns identified in Task 3 where 0 is an implausible value
columns_to_impute = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']

# Replace 0 with NaN in the identified columns
# .replace(0, np.nan) is efficient for this operation
df[columns_to_impute] = df[columns_to_impute].replace(0, np.nan)
print("Replaced implausible zero values with NaN.\n")

# Verify the number of NaN values after replacement
print("NaN values after replacing zeros:")
print(df[columns_to_impute].isnull().sum())

print("\n--- Imputing Missing Values with Median ---\n")

# Impute missing values (NaN) with the median of each column
filled_counts = {}
for col in columns_to_impute:
    # Calculate median *before* imputation to report it
    median_val = df[col].median()

    # Count NaNs before filling to report how many were filled
    na_before = df[col].isnull().sum()

    # Fill NaN values with the median of the column
    df[col] = df[col].fillna(median_val) # Modified: Assign back the result to avoid FutureWarning

    # Store the count of filled values for reporting
    filled_counts[col] = na_before
    print(f"Column '{col}': Filled {na_before} missing values with median ({median_val:.2f}).")

print("\n--- Verification After Cleaning ---\n")
print("DataFrame Info after cleaning and imputation:")
df.info()

print("\nDescriptive Statistics after cleaning and imputation:")
display(df.describe())
```
